# Beginning of Activity: Notebook 1 - EDA

```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
from statsmodels.tsa.stattools import adfuller
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf       
```

```python
# Load the dataset
data_1 = pd.read_excel('ISBN List.xlsx', sheet_name=None)  
data_2 = pd.read_excel('UK Weekly Trended Timeline from 200101_202429.xlsx', sheet_name=None)
```

```python
print(f"File 1 sheets: {list(data_1.keys())}")
print(f"File 2 sheets: {list(data_2.keys())}")
```

```python
# Combine all sheets in data_1 into a single DataFrame with an extra column to identify the sheet
df_isbn = pd.concat(
    [df.assign(Category=sheet_name) for sheet_name, df in data_1.items()],
    ignore_index=True
)
```

```python
# Combine all sheets in data_2 into a single DataFrame with an extra column to identify the sheet
df_uk_weekly_trend = pd.concat(
    [df.assign(Category=sheet_name) for sheet_name, df in data_2.items()],
    ignore_index=True
)
```

```python
display(df_isbn.head(5), df_uk_weekly_trend.head(5))  # Display the first few rows of the combined ISBN and UK weekly trend DataFrames
```

```python
#replace spaces with underscores across all column names
df_uk_weekly_trend.columns = df_uk_weekly_trend.columns.str.replace(' ', '_')
df_isbn.columns = df_isbn.columns.str.replace(' ', '_')
```

```python
# Identify date columns in both DataFrames
date_col_df_isbn = [col for col in df_isbn.columns if pd.api.types.is_datetime64_any_dtype(df_isbn[col])]
date_col_df_uk_trend = [col for col in df_uk_weekly_trend.columns if pd.api.types.is_datetime64_any_dtype(df_uk_weekly_trend[col])]
print("Date columns found:", date_col_df_isbn)
print("Date columns found:", date_col_df_uk_trend)
```

```python
# Set datetime columns as index for each dataframe and sort by index
df_isbn['Publication_Date'] = pd.to_datetime(df_isbn['Publication_Date'], errors='coerce')
df_isbn.set_index(df_isbn['Publication_Date'], inplace=True)
df_isbn.sort_index(inplace=True)  # Sort by index
# drop the 'Publication_Date' column as it's now the index
df_isbn.drop(columns=['Publication_Date'], inplace=True)

df_uk_weekly_trend['End_Date'] = pd.to_datetime(df_uk_weekly_trend['End_Date'], errors='coerce')
df_uk_weekly_trend.set_index(df_uk_weekly_trend['End_Date'], inplace=True)  # Use the first datetime column as index
df_uk_weekly_trend.sort_index(inplace=True)  # Sort by index
# drop the 'End_Date' column as it's now the index
df_uk_weekly_trend.drop(columns=['End_Date'], inplace=True)
```

```python
# Convert ISBN columns to string: We need to keep this information as is.
df_isbn['ISBN'] = df_isbn['ISBN'].astype(str)
df_uk_weekly_trend['ISBN'] = df_uk_weekly_trend['ISBN'].astype(str)
```

```python
display(df_isbn.tail(2), df_uk_weekly_trend.tail(2))  # Display the last few rows of each DataFrame to confirm sorting
```

Note: df_isbn contains data only until 2023-05-23. This means that this data set will not have sales data after applying the filtering criteria estipulated in the project context.

```python
print(df_isbn.shape, df_uk_weekly_trend.shape)  # Print the shape of both DataFrames to confirm they are not empty
```

### Resample the data to weekly frequency and fill in missing weeks with zeros


```python
# Resample by week (W), sum the data, and fill missing values with 0, modifying the original dataframe. Reshape inplace, otherwise the ISBNs are not preserved.
df_isbn.resample('W').sum().fillna(0, inplace=True)
df_uk_weekly_trend.resample('W').sum().fillna(0, inplace=True)
```

```python
print(df_isbn.shape, df_uk_weekly_trend.shape)  # Print the shape of the reshaped DataFrames
```

```python
display(df_uk_weekly_trend.head(5))  # Display the first few rows of the reshaped DataFrames
```

### Visual examination of the sales patterns for the selected ISBNs.

```python
# Create a plot grid (e.g., 8x8 grid for 64 slots, leaving some empty space for 61 ISBNs)
isbn_filtered = uk_trend_weekly_filtered['ISBN'].unique()

num_plots = len(isbn_filtered)
rows = (num_plots // 6) + (1 if num_plots % 6 != 0 else 0)  # Calculate number of rows needed
cols = 6  # 8 plots per row
```

```python
# Create the figure and axes for subplots
fig, axes = plt.subplots(rows, cols, figsize=(20, 4 * rows))

# Flatten the axes array for easy indexing
axes = axes.flatten()

# Loop through each ISBN and plot the data from the original df_uk_weekly_trend dataframe
for idx, isbn in enumerate(isbn_filtered):
    # Filter the original df_uk_weekly_trend to get data for the current ISBN
    df_uk_weekly_isbn = df_uk_weekly_trend[df_uk_weekly_trend['ISBN'] == isbn]

    # Plot the data
    axes[idx].plot(df_uk_weekly_isbn.index, df_uk_weekly_isbn['Volume'], label=f'ISBN: {isbn}')
    axes[idx].set_title(f'ISBN: {isbn}')
    axes[idx].set_xlabel('Date')
    axes[idx].set_ylabel('Volume')
    axes[idx].legend()

# Remove unused subplots if there are any
for idx in range(num_plots, len(axes)):
    fig.delaxes(axes[idx])

# Adjust layout for better spacing
plt.tight_layout()
```

Note: Most of ISBNs (Books) sold during this period of time present seasonal patterns and a strong negative trend. This may be explained by the fact that these books have been in the market for a long period of time. Some books show strong peaks that fluctuate in what it seems to be seasonal patterns. Interestingly, one book (ISBN: 9780241003008) shows a more neutral trend with a seasonal or cyclical pattern that much longer than the other books, of approximately 12 years.
